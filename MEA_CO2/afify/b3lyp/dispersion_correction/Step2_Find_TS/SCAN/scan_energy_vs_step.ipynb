```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# load data
data = np.loadtxt("orca.relaxscanact.dat")
df = pd.DataFrame(data, columns=["Coordinate", "Energy_Ha"])
df["Step"] = np.arange(1, len(df) + 1)

# relative energy (kcal/mol)
HARTREE_TO_KCALMOL = 627.509474
df["RelE_kcalmol"] = (df["Energy_Ha"] - df["Energy_Ha"].min()) * HARTREE_TO_KCALMOL

print(f"N steps: {len(df)}")
print(f"Energy min (Ha): {df.Energy_Ha.min():.8f}")
print(f"Energy max (Ha): {df.Energy_Ha.max():.8f}")
print(
    f"Energy range (kcal/mol): {(df.Energy_Ha.max()-df.Energy_Ha.min())*HARTREE_TO_KCALMOL:.2f}"
)
df.head()
```

```python
# plot: step vs energy
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df["Step"], df["RelE_kcalmol"], "o-", markersize=4, linewidth=1.8)
ax.set_xlabel("Step")
ax.set_ylabel("Relative Energy (kcal/mol)")
ax.set_title("Relaxed Surface Scan: Energy vs Step")
ax.grid(True, alpha=0.3, linestyle="--")
plt.tight_layout()
plt.savefig("scan_energy_vs_step.png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: scan_energy_vs_step.png")
```
