## 化学的解釈

### 1. 遷移状態の確定

✓ **虚数周波数 1つ** = 一次の鞍点 = 遷移状態

### 2. 反応座標

虚数周波数 (-202.42 cm⁻¹) に対応するノーマルモードが**反応座標**を表す

典型的に、MEA-CO₂結合形成反応では：
- **N-CO₂ C距離の短縮** （or 伸長）
- **CO₂の向き** の変化

### 3. エネルギーバリア

- **活性化エネルギー**: ~12.8 kJ/mol
- **反応物（Step 1）** → **TS（Step 13）** → **生成物**

バリアが小さい（~12 kJ/mol）ため：
- **常温反応が容易**
- **可逆性の可能性** (forward/backward)

## 主要な原子間距離（TS構造）

```
N-CO₂ C: 2.052 Å  ← 形成中の新結合
N-水 O: 2.573 Å  ← 水素結合相互作用
MEA C-C: 1.516 Å  ← MEA骨格
```

**構造的特徴**:
- N原子が CO₂ と適切な距離で相互作用
- 水分子が周辺に配置
- MEA 側鎖は比較的硬い（ほぼ不変）
